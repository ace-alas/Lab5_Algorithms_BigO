```python
# [CELL 2] STUDENT CONFIGURATION (student-specific input rules)
LAST_NAME = "Siruno"
STUDENT_ID = "TUPM-26-7958"
FAVORITE_ARTIST = "BINI"   # <-- CHANGE THIS to YOUR favorite artist

SEED_DIGIT = int(STUDENT_ID[-1])                                  # last digit of ID
ID_SUM = sum(int(d) for d in STUDENT_ID if d.isdigit())          # sum of ID digits
NAME_LENGTH = len(LAST_NAME)                                      # length of surname

_artist_letters = [c for c in FAVORITE_ARTIST.upper() if c.isalpha()]
ARTIST_LENGTH = len(_artist_letters)                              # letters only (spaces ignored)
ARTIST_SUM = sum(ord(c) - ord("A") + 1 for c in _artist_letters)  # A=1 ... Z=26

print("LAST_NAME      :", LAST_NAME)
print("STUDENT_ID     :", STUDENT_ID)
print("FAVORITE_ARTIST:", FAVORITE_ARTIST)
print("SEED_DIGIT     :", SEED_DIGIT)
print("ID_SUM         :", ID_SUM)
print("NAME_LENGTH    :", NAME_LENGTH)
print("ARTIST_LENGTH  :", ARTIST_LENGTH)
print("ARTIST_SUM     :", ARTIST_SUM)
```

```python
# [CELL 3] SORTING ALGORITHMS  (each returns: sorted list, comparisons, movements)
def bubble_sort(data):
    a, comparisons, swaps = data.copy(), 0, 0
    for i in range(len(a)):
        swapped = False
        for j in range(len(a) - i - 1):
            comparisons += 1
            if a[j] > a[j + 1]:
                a[j], a[j + 1] = a[j + 1], a[j]
                swaps += 1
                swapped = True
        if not swapped:
            break
    return a, comparisons, swaps


def selection_sort(data):
    a, comparisons, swaps = data.copy(), 0, 0
    for i in range(len(a) - 1):
        minimum = i
        for j in range(i + 1, len(a)):
            comparisons += 1
            if a[j] < a[minimum]:
                minimum = j
        if minimum != i:
            a[i], a[minimum] = a[minimum], a[i]
            swaps += 1
    return a, comparisons, swaps


def insertion_sort(data):
    a, comparisons, shifts = data.copy(), 0, 0
    for i in range(1, len(a)):
        key, j = a[i], i - 1
        while j >= 0:
            comparisons += 1
            if a[j] <= key:
                break
            a[j + 1] = a[j]
            shifts += 1
            j -= 1
        a[j + 1] = key
    return a, comparisons, shifts
```

```python
# [CELL 4] GENERATE ORDERS AND THE THREE INPUT CONDITIONS (same orders)
import random
import time

data_size = 100 + ID_SUM * 5
orders = [(ID_SUM * 3 + i * 7) % 1000 for i in range(data_size)]   # unique order numbers

rng = random.Random(SEED_DIGIT * ID_SUM)        # fixed seed = reproducible results
random_orders = orders.copy()
rng.shuffle(random_orders)

conditions = {
    "Sorted":  sorted(orders),
    "Reverse": sorted(orders, reverse=True),
    "Random":  random_orders,
}

print("Dataset Size:", data_size)
print("Generated orders (first 15):", orders[:15])
for name, values in conditions.items():
    print(f"{name:<8} input (first 10):", values[:10])
```

```python
# [CELL 5] EXECUTION-TIME MEASUREMENT
def measure(sort_function, data, trials=5):
    times = []
    for _ in range(trials):
        start = time.perf_counter()
        result = sort_function(data)
        times.append(time.perf_counter() - start)
    return result, sum(times) / trials
```

```python
# [CELL 6] RUN ALL ALGORITHMS ON ALL CONDITIONS + VERIFY
algorithms = [
    ("Bubble Sort", bubble_sort),
    ("Selection Sort", selection_sort),
    ("Insertion Sort", insertion_sort),
]
expected = sorted(orders)
results = []   # (condition, algorithm, comparisons, movements, time, verified)

for cond_name, cond_data in conditions.items():
    for algo_name, algo in algorithms:
        (result, comps, moves), elapsed = measure(algo, cond_data)
        results.append((cond_name, algo_name, comps, moves, elapsed, result == expected))
```

```python
# [CELL 7] SORTING PERFORMANCE COMPARISON
print(f"\n{'Condition':<10}{'Algorithm':<18}{'Comparisons':>12}{'Movements':>12}"
      f"{'Time (s)':>14}{'Sorted?':>9}")
print("-" * 75)
for cond, name, comps, moves, elapsed, ok in results:
    print(f"{cond:<10}{name:<18}{comps:>12}{moves:>12}{elapsed:>14.8f}{str(ok):>9}")

print("\nAll results correctly sorted:", all(r[5] for r in results))
```
